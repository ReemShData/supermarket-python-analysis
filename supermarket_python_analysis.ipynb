```python
import pandas as pd

df = pd.read_csv('/kaggle/input/datasets/reemalmutairi1/supermarket-clean/supermarket_clean.csv')
print(df.shape)
print(df.head())
```

```python
# Step 2: Total Profit by Branch
branch_profit = df.groupby('Branch')['Profit'].sum().round(2).sort_values(ascending=False)
print(branch_profit)
```

```python
# Step 3: Total Profit by Category
category_profit = df.groupby('Category')['Profit'].sum().round(2).sort_values(ascending=False)
print(category_profit)
```

```python
# Step 4: Total Profit by Payment Type
payment_profit = df.groupby('Payment_Type')['Profit'].sum().round(2).sort_values(ascending=False)
print(payment_profit)
```

```python
# Step 5: Total Profit by Month
df['Month'] = pd.to_datetime(df['Order_Date']).dt.month
month_profit = df.groupby('Month')['Profit'].sum().round(2).sort_values(ascending=False)
print(month_profit)
```

```python
# Step 6: Bar Chart - Total Profit by Branch
import matplotlib.pyplot as plt

branch_profit.plot(kind='bar', color='steelblue', figsize=(10,5))
plt.title('Total Profit by Branch')
plt.xlabel('Branch')
plt.ylabel('Total Profit ($)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Step 7: Bar Chart - Total Profit by Category
category_profit.plot(kind='bar', color='coral', figsize=(10,5))
plt.title('Total Profit by Category')
plt.xlabel('Category')
plt.ylabel('Total Profit ($)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Step 8: Pie Chart - Payment Type Distribution
payment_profit.plot(kind='pie', autopct='%1.1f%%', figsize=(5,5))
plt.title('Payment Type Distribution')
plt.ylabel('')
plt.show()
```

```python
# Step 9: Line Chart - Total Profit by Month
month_profit.sort_index().plot(kind='line', marker='o', color='green', figsize=(10,5))
plt.title('Total Profit by Month')
plt.xlabel('Month')
plt.ylabel('Total Profit ($)')
plt.xticks(range(1,13))
plt.tight_layout()
plt.show()
```
